```python
import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = str(text).lower()                          
    text = re.sub(r'[^a-z\s]', '', text)               
    text = ' '.join(word for word in text.split() if word not in stop_words)  
    return text

data['clean_text'] = (data['Ticket Subject'] + ' ' + data['Ticket Description']).apply(clean_text)
data['clean_text']

```

```python
custom_stopwords = stop_words.union({'productpurchased', 'im'})

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z\s]', '', text)
    text = ' '.join(word for word in text.split() if word not in custom_stopwords)
    return text

data['clean_text'] = (data['Ticket Subject'] + ' ' + data['Ticket Description']).apply(clean_text)
data['clean_text']
```

```python
data['clean_text'].head(10)
```

```python
custom_stopwords = stop_words.union({'productpurchased', 'im', 'please', 'assist', 'issue'})
custom_stopwords
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=3000)
X = vectorizer.fit_transform(data['clean_text'])

print(X.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

y_type = data['Ticket Type']

X_train, X_test, y_train, y_test = train_test_split(
    X, y_type, test_size=0.2, random_state=42, stratify=y_type
)

model_type = LogisticRegression(max_iter=1000)
model_type.fit(X_train, y_train)

y_pred = model_type.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
y_priority = data['Ticket Priority']

X_train, X_test, y_train, y_test = train_test_split(
    X, y_priority, test_size=0.2, random_state=42, stratify=y_priority
)

model_priority = LogisticRegression(max_iter=1000)
model_priority.fit(X_train, y_train)

y_pred = model_priority.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred, labels=model_priority.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model_priority.classes_)
disp.plot(cmap='Blues')
plt.title('Priority Prediction Confusion Matrix')
plt.show()
```

```python
'''Approach: TF-IDF (3000 features) + Logistic Regression, trained separately for Ticket Type (5 classes) and Ticket Priority (4 classes), on the Kaggle Customer Support Ticket Dataset.
Results: ~20% accuracy for category (vs. 20% random baseline), ~24% accuracy for priority (vs. 25% random baseline). Include both classification reports and this confusion matrix image.
Analysis: Both models perform at chance level. The confusion matrix shows uniformly distributed errors with no diagonal concentration, indicating the model found no learnable relationship between ticket text and either label.
Conclusion: This points to a limitation in the dataset itself — the text field appears synthetically/randomly generated and not causally linked to the assigned category or priority labels, unlike real-world tickets where an 
agent's actual reasoning drives both the text and the label.
What would fix it: A dataset with real, human-written tickets and human-assigned labels (e.g. the Zenodo dataset also listed in the task) would likely show genuine text-to-label correlation, since real support text usually does 
telegraph both category and urgency in its wording.'''
```
